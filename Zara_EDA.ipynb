```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("Zara_Sales_Analysis.csv", encoding= 'latin1')
```

```python
df['name'] = df['name'].fillna('count')
df['description'] = df['description'].fillna('count')
```

```python
df.info()
```

```python
df.corr(numeric_only=True)
```

```python
df['revenue']= df['price'] * df['sales_volume']
```

```python
df.columns
```

```python
###Top Selling Products
top_products = df.sort_values(by='sales_volume', ascending=False).head(5)

bars = plt.barh(top_products['name'], top_products['sales_volume'])

plt.title("Top Selling Products")
plt.xlabel("Sales Volume")

# Bar ke aage numbers
for bar in bars:
    width = bar.get_width()
    plt.text(
        width,
        bar.get_y() + bar.get_height()/2,
        int(width),
        va='center'
    )

plt.show()
```

```python
##Section-wise Sales Analysis
df.groupby('section')['sales_volume'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("Section Wise Sales")
plt.xlabel("Section")
plt.ylabel("sales_volume")
plt.show()
```

```python
##Promotion Impact Analysis
df.groupby('promotion')['revenue'].sum().plot(kind='pie', autopct='%1.1f%%')
plt.ylabel('')
plt.title("Promotion Impact Analysis")
plt.show()
```

```python
##Seasonal Product Analysis
df.groupby('seasonal')['revenue'].mean().plot(kind='bar')
plt.title('Seasonal Product Analysis')
plt.show()
```

```python
##Price vs Sales Relationship
plt.scatter(df['price'], df['sales_volume'])

plt.xlabel("Price")
plt.ylabel("Sales Volume")
plt.title("Price vs Sales")
plt.show()
```

```python
##Product Position Analysis
df.groupby('product_position\n')['sales_volume'].mean().plot(kind='bar')

plt.title("Product Position Impact")
plt.show()
```
